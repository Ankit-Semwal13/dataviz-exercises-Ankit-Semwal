```python
# Task 1 — Multi-series line with highlight

asia = df[df["Region"] == "Asia"].copy()
highlight_country = "China"

fig = go.Figure()

for country in asia["Country"].unique():
    country_df = asia[asia["Country"] == country]

    if country == highlight_country:
        fig.add_trace(go.Scatter(
            x=country_df["Year"],
            y=country_df["CO2_Mt"],
            mode="lines",
            line=dict(color="#D62728", width=4),
            name=country,
            showlegend=False
        ))
    else:
        fig.add_trace(go.Scatter(
            x=country_df["Year"],
            y=country_df["CO2_Mt"],
            mode="lines",
            line=dict(color="#DDDDDD", width=1),
            name=country,
            showlegend=False
        ))

# Direct label at the end of highlighted line
highlight_df = asia[asia["Country"] == highlight_country]
last_point = highlight_df[highlight_df["Year"] == highlight_df["Year"].max()]

fig.add_annotation(
    x=last_point["Year"].iloc[0],
    y=last_point["CO2_Mt"].iloc[0],
    text=highlight_country,
    showarrow=False,
    xshift=35,
    font=dict(size=14, color="#D62728", family="Arial")
)

fig.update_layout(
    title="China’s CO2 emissions rose far above other Asian countries",
    title_font=dict(size=22, family="Arial"),
    font=dict(family="Arial", size=13),
    plot_bgcolor="white",
    paper_bgcolor="white",
    xaxis_title="Year",
    yaxis_title="CO2 emissions, Mt",
    margin=dict(l=60, r=100, t=80, b=60)
)

fig.update_xaxes(showgrid=False, linecolor="black")
fig.update_yaxes(showgrid=True, gridcolor="#EEEEEE", linecolor="black")

fig.show()
```
